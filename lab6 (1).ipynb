```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Cars Datasets 2025.csv', encoding='latin1')

print("Первые строки датасета:")
display(df.head())

print("\nИнформация о датасете:")
df.info()

print("\nКоличество строк и столбцов:", df.shape)
print("\nСписок столбцов:", df.columns.tolist())
print("\nТипы данных:")
print(df.dtypes)

print("\nДоли пропусков по столбцам:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
```

```python
df['HorsePower'] = df['HorsePower'].astype(str)
for char in [' hp']:
  df['HorsePower'] = df['HorsePower'].str.replace(char, '', regex=False)

df['Total Speed'] = df['Total Speed'].astype(str)
for char in [' km/h']:
  df['Total Speed'] = df['Total Speed'].str.replace(char, '', regex=False)

df['Performance(0 - 100 )KM/H'] = df['Performance(0 - 100 )KM/H'].astype(str)
for char in [' sec']:
  df['Performance(0 - 100 )KM/H'] = df['Performance(0 - 100 )KM/H'].str.replace(char, '', regex=False)

df['Cars Prices'] = df['Cars Prices'].astype(str)
for char in ['$', ',', ' ']:
  df['Cars Prices'] = df['Cars Prices'].str.replace(char, '', regex=False)

df['CC/Battery Capacity'] = df['CC/Battery Capacity'].astype(str)
for char in [',', ' cc']:
  df['CC/Battery Capacity'] = df['CC/Battery Capacity'].str.replace(char, '', regex=False)

df['Torque'] = df['Torque'].astype(str)
for char in [' Nm']:
  df['Torque'] = df['Torque'].str.replace(char, '', regex=False)

df['Seats'] = df['Seats'].astype(str)
df['Seats'] = pd.to_numeric(df['Seats'], errors='coerce')

print("Датасет после очистки:")
display(df.head())

print("\nТипы данных после очистки:")
print(df.dtypes)

print("\nДоли пропусков после очистки:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
df['CC/Battery Capacity'] = pd.to_numeric(df['CC/Battery Capacity'], errors='coerce')
df['HorsePower'] = pd.to_numeric(df['HorsePower'], errors='coerce')
df['Total Speed'] = pd.to_numeric(df['Total Speed'], errors='coerce')
df['Performance(0 - 100 )KM/H'] = pd.to_numeric(df['Performance(0 - 100 )KM/H'], errors='coerce')
df['Cars Prices'] = pd.to_numeric(df['Cars Prices'], errors='coerce')
df['Torque'] = pd.to_numeric(df['Torque'], errors='coerce')

df.drop_duplicates(inplace=True)

print("Количество строк и столбцов после удаления дубликатов:", df.shape)
```

```python
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype in ['float64', 'int64']:
            median_val = df[col].median()
            df[col].fillna(median_val, inplace=True)
        elif df[col].dtype == 'object':
            df[col].fillna("Unknown", inplace=True)

print("\nДоли пропусков после заполнения:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64'])
descriptive_stats = numeric_cols.agg(['min', 'max', 'mean', 'median'])
display(descriptive_stats)
```

```python
print("Самая дорогая модель:")
display(df.loc[df['Cars Prices'].idxmax()][['Company Names', 'Cars Names', 'Cars Prices']])

print("\nСамая быстрая модель автомобиля (по разгону до 100 км/ч):")
display(df.loc[df['Performance(0 - 100 )KM/H'].idxmin()][['Company Names', 'Cars Names', 'Performance(0 - 100 )KM/H']])

print("\nСамая мощная модель автомобиля:")
display(df.loc[df['HorsePower'].idxmax()][['Company Names', 'Cars Names', 'HorsePower']])
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(18, 6))

axes[0].hist(df['Cars Prices'], bins=50, color='skyblue')
axes[0].set_xlabel('Цена')
axes[0].set_title('Цены')

axes[1].hist(df['HorsePower'], bins=50, color='lightcoral')
axes[1].set_xlabel('Лошадиные силы')
axes[1].set_title('Лошадиные силы')

axes[2].hist(df['Total Speed'], bins=50, color='lightgreen')
axes[2].set_xlabel('Скорость')
axes[2].set_title('Скорость')

fig.suptitle('Изменения', y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

```python
top10 = df['Company Names'].value_counts().head(10)

plt.figure(figsize=(10, 10))
plt.pie(top10, labels=top10.index, autopct='%1.1f%%', startangle=140)
plt.title('Топ 10 Брендов по кол-ву моделей')
plt.axis('equal')
plt.show()
```

```python
dftop10 = df[df['Company Names'].isin(top10.index)].copy()

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(20, 8))

axes[0].boxplot([dftop10[dftop10['Company Names'] == company]['Cars Prices'].dropna() for company in top10.index], labels=top10.index)
axes[0].set_title('Цены машин для топ10')
axes[0].set_xlabel('Компания')
axes[0].set_ylabel('Цены')
axes[0].tick_params(axis='x', rotation=45)

axes[1].boxplot([dftop10[dftop10['Company Names'] == company]['HorsePower'].dropna() for company in top10.index], labels=top10.index)
axes[1].set_title('ЛС для топ10')
axes[1].set_xlabel('Компания')
axes[1].set_ylabel('ЛС')
axes[1].tick_params(axis='x', rotation=45)

axes[2].boxplot([dftop10[dftop10['Company Names'] == company]['Total Speed'].dropna() for company in top10.index], labels=top10.index)
axes[2].set_title('Макс скорость для топ10')
axes[2].set_xlabel('Компания')
axes[2].set_ylabel('Макс скорость')
axes[2].tick_params(axis='x', rotation=45)


fig.suptitle('Топ10 брендов', y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

```python
matrix = df.select_dtypes(include=['float64', 'int64']).corr(numeric_only=True)

plt.figure(figsize=(10, 8))
sns.heatmap(matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.show()

price = matrix['Cars Prices'].drop('Cars Prices')
top3 = price.abs().sort_values(ascending=False).head(3)

print("\nТоп 3 самые сильные связи с ценой:")
for col in top3.index:
    print(f"- {col}: {price[col]:.2f}")
```
